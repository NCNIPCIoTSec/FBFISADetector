```python
import os
from IPython.display import display, SVG

from settings import OUTPUT_DIR

for arch in os.listdir(OUTPUT_DIR):
    if not os.path.isdir(f'{OUTPUT_DIR}{arch}'):
        continue
    for out_file in os.listdir(f"{OUTPUT_DIR}{arch}"):
        if not out_file.endswith(".svg"):
            continue
        file_path = f"{OUTPUT_DIR}{arch}/{out_file}"
        print(file_path)
        display(SVG(file_path))
```

```python
import matplotlib.pyplot as plt

import settings
from evaluate.lab.ablationstudy import create_segmented_histogram, ablation_study


data = ablation_study(samples_workspace=settings.SAMPLES_DIR, output_workspace=settings.OUTPUT_DIR)

# 定义分段参数
bins = 4  # 直方图分箱数量
segments = 2  # 每个bin要分割的段数
colors = ['#666666', '#FFFFFF']  # 每段使用的颜色

# 创建分段直方图
fig, ax = create_segmented_histogram(
    data=data, 
    bin_num=bins, 
    bin_tags=['A', 'B', 'C', 'D'],
    segments=segments, 
    seg_tags=['Valid', 'InValid'],
    seg_colors=colors,
    title="Lab Display",
    x_label="Type",
    y_label="Frequency",
    hist_hw_rate=0.03
)

plt.show()  
```
